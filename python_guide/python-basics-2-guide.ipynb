## Strings: Indexing and Slicing

A string is a sequence of characters. You can read a single character by its position (index), and a range of characters with a slice `text[start:stop:step]`. Indexing starts at `0`, and `-1` is the last character. The `stop` position is **not** included.

```python
text = "Python"

print(text[0])    # P  (first character)
print(text[-1])   # n  (last character)

print(text[0:3])  # Pyt  (positions 0, 1, 2)
print(text[:3])   # Pyt  (from the start)
print(text[3:])   # hon  (to the end)
print(text[::-1]) # nohtyP  (reversed)
print(len(text))  # 6
```

```python
name = "She is a girl"

print(name[:2])
print(len(name))

print(name[::-1])
```

## More String Methods

Strings come with many helpful methods. They return a **new** string (strings cannot be changed in place).

```python
raw = "   Hello, World   "

print(raw.strip())              # "Hello, World" (trim spaces)
print("Hello".startswith("He")) # True
print("file.txt".endswith(".txt"))  # True
print("banana".count("a"))      # 3
print("Hello, World".find("World"))  # 7 (index where it starts, -1 if missing)

# split and join are opposites of each other
parts = "a,b,c".split(",")      # ['a', 'b', 'c']
print(parts)
print("-".join(parts))          # "a-b-c"

print("Hello".isalpha())        # True
print("123".isdigit())          # True

print(name)
```

```python
x = "   Hello, Everyone       "

print(x.strip())
print(x.count("e"))
```

## Formatting with f-strings

f-strings can format numbers and align text. Put a format specifier after a colon inside the braces.

| Spec | Meaning | Example |
| --- | --- | --- |
| `:.2f` | 2 decimal places | `3.14` |
| `:,` | thousands separator | `1,000,000` |
| `:>8` | right-align in 8 spaces | `   hello` |
| `:<8` | left-align in 8 spaces | `hello   ` |
| `:.1%` | percentage | `25.0%` |

```python
price = 1234.5678
print(f"{price:.2f}")     # 1234.57
print(f"{price:,.2f}")    # 1,234.57

ratio = 0.25
print(f"{ratio:.1%}")     # 25.0%

name = "Alice"
print(f"|{name:<8}|")     # |Alice   |
print(f"|{name:>8}|")     # |   Alice|

# Build a small aligned table
for item, qty in [("apples", 3), ("rice", 12), ("milk", 1)]:
    print(f"{item:<10} {qty:>3}")
```

```python
for name , age in [("Sohail", 26),("Imran" , 30),("Rexa" , 35)]:
    print (name , age)


z = 124315.857
print(f"{z:.1f}")
print(f"{z:,.3f}")
```

## Looping with enumerate

When you need both the position **and** the value while looping, use `enumerate` instead of a manual counter.

```python
fruits = ["apple", "banana", "cherry"]

# Without enumerate (works, but clunky)
index = 0
for fruit in fruits:
    print(index, fruit)
    index += 1

print("---")

# With enumerate (preferred)
for i, fruit in enumerate(fruits):
    print(i, fruit)

# Start counting from 1
for i, fruit in enumerate(fruits, start=1):
    print(f"{i}. {fruit}")
```

```python
car = ["VW", "Skoda","BMW","Tata"]

for i, car in enumerate(car):
    print(i, car)

fruit = ["mango", "raspberry", "chikoo"]

for i, fruit in enumerate(fruit, start=2):
    print(f"{i} . {fruit}")
```

## Looping over two lists with zip

`zip` pairs up items from several sequences so you can loop over them together. It stops at the shortest sequence.

```python
names = ["Alice", "Bob", "Carol"]
scores = [85, 92, 78]

for name, score in zip(names, scores):
    print(f"{name}: {score}")

# zip is handy for building a dictionary
lookup = dict(zip(names, scores))
print(lookup)   # {'Alice': 85, 'Bob': 92, 'Carol': 78}
```

```python
car = ["Polo", "Octavia","Duster"]
brand = ["VW","Skoda","renault"]

for car, brand in zip(car,brand):
    print(f"{car} : {brand}")
```

## Sorting with a Key

`sorted()` returns a new sorted list; `list.sort()` sorts in place. Pass a `key` function to decide what to sort by, and `reverse=True` for descending order.

```python
words = ["banana", "kiwi", "apple", "fig"]

print(sorted(words))               # alphabetical
print(sorted(words, key=len))      # by length: ['kiwi', 'fig'...] shortest first
print(sorted(words, reverse=True)) # reverse alphabetical

# Sort a list of (name, score) pairs by score, highest first
scores = [("Alice", 85), ("Bob", 92), ("Carol", 78)]
ranked = sorted(scores, key=lambda pair: pair[1], reverse=True)
print(ranked)
```

## Nested Lists

A list can contain other lists. This is useful for grids, tables, or rows of data. Use two indexes to reach an inner item: `grid[row][column]`.

```python
grid = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print(grid[0])      # [1, 2, 3]  (first row)
print(grid[0][2])   # 3          (first row, third column)

# Loop over every value
for row in grid:
    for value in row:
        print(value, end=" ")
    print()   # newline after each row
```

```python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9],
]

print(matrix)

for x in matrix:
    for y in x:
        print(y, end=" ")
    print()
```

## Copying Lists (a Common Gotcha)

Assigning a list to a new name does **not** copy it — both names point to the same list. To make an independent copy, use `.copy()` or `list(...)`.

```python
original = [1, 2, 3]

alias = original          # NOT a copy: same list
alias.append(4)
print(original)           # [1, 2, 3, 4]  -- original changed too!

# A real copy
real_copy = original.copy()
real_copy.append(99)
print(original)           # [1, 2, 3, 4]  -- unchanged
print(real_copy)          # [1, 2, 3, 4, 99]
```

## A Gentle Intro to List Comprehensions

A list comprehension builds a new list in one line. It is a shorter way to write a simple `for` loop that collects results. You will use these constantly.

`[expression for item in sequence]` — and you can add `if condition` to filter.

```python
# The long way
squares = []
for n in range(6):
    squares.append(n * n)
print(squares)   # [0, 1, 4, 9, 16, 25]

# The same thing as a comprehension
squares = [n * n for n in range(6)]
print(squares)

# With a filter: keep only even numbers
evens = [n for n in range(10) if n % 2 == 0]
print(evens)     # [0, 2, 4, 6, 8]

# Transform a list of strings
names = ["alice", "bob"]
upper = [name.upper() for name in names]
print(upper)     # ['ALICE', 'BOB']
```

## Dictionaries: Safer Access

Reading a missing key with `d[key]` raises an error. Use `.get(key, default)` to provide a fallback, and `.setdefault(key, default)` to insert a default only if the key is missing.

```python
person = {"name": "Alice", "age": 25}

print(person.get("email"))                 # None (no error)
print(person.get("email", "not set"))      # not set

# setdefault inserts the key if it is missing, then returns the value
person.setdefault("city", "London")
person.setdefault("name", "Bob")           # name already exists, so ignored
print(person)
```

## Counting with a Dictionary

A very common pattern: count how many times each item appears.

```python
votes = ["yes", "no", "yes", "yes", "no"]

counts = {}
for vote in votes:
    counts[vote] = counts.get(vote, 0) + 1

print(counts)   # {'yes': 3, 'no': 2}

# Loop over the results
for option, total in counts.items():
    print(f"{option}: {total}")
```

## Nested Data: Lists of Dictionaries

Real data is often a list of records, where each record is a dictionary. This is exactly the shape of data you get from spreadsheets, APIs, and databases.

```python
people = [
    {"name": "Alice", "age": 30, "city": "London"},
    {"name": "Bob", "age": 25, "city": "Paris"},
    {"name": "Carol", "age": 35, "city": "London"},
]

# Print everyone's name and city
for person in people:
    print(f"{person['name']} lives in {person['city']}")

# Find the average age
ages = [person["age"] for person in people]
print("Average age:", sum(ages) / len(ages))

# Filter: only people in London
londoners = [p["name"] for p in people if p["city"] == "London"]
print("In London:", londoners)
```

## Functions That Return Several Values

A function can return more than one value by separating them with commas. Python packs them into a tuple, which you can unpack on the receiving side.

```python
def min_and_max(numbers):
    return min(numbers), max(numbers)

low, high = min_and_max([4, 8, 1, 9, 3])
print(low, high)   # 1 9

def divide(a, b):
    quotient = a // b
    remainder = a % b
    return quotient, remainder

q, r = divide(17, 5)
print(f"{q} remainder {r}")   # 3 remainder 2
```

## Keyword Arguments and Defaults

You can pass arguments by name, which makes calls easier to read. Parameters with defaults must come **after** parameters without defaults.

```python
def make_user(name, role="member", active=True):
    return {"name": name, "role": role, "active": active}

print(make_user("Alice"))
print(make_user("Bob", role="admin"))
print(make_user("Carol", active=False))

# Naming arguments makes the call clear, in any order
print(make_user(name="Dave", active=False, role="guest"))
```

## Variable Scope: Local vs Global

A variable created **inside** a function is *local* — it only exists there. Code inside a function can read variables from the outside, but assigning to a name creates a new local one unless you say otherwise.

Prefer passing values in as arguments and returning results rather than relying on global variables.

```python
message = "global"   # defined outside

def show():
    message = "local"   # a separate, local variable
    print("inside:", message)

show()
print("outside:", message)   # still "global"

# A function can READ an outer variable without reassigning it
tax_rate = 0.2
def add_tax(price):
    return price * (1 + tax_rate)   # reads the global tax_rate

print(add_tax(100))   # 120.00
```

## Docstrings

A docstring is a string at the top of a function that explains what it does. Tools and the built-in `help()` use it. Good docstrings make your code easier to use.

```python
def celsius_to_fahrenheit(celsius):
    """Convert a temperature from Celsius to Fahrenheit."""
    return celsius * 9 / 5 + 32

print(celsius_to_fahrenheit(100))   # 212.00 # it will give 2 decimal values
help(celsius_to_fahrenheit)
```

## Truthiness

In a condition, values are treated as `True` or `False`. Empty things (`0`, `""`, `[]`, `{}`, `None`) are "falsy"; most other values are "truthy". This lets you write clean checks.

```python
items = []

if items:
    print("There are items")
else:
    print("The list is empty")   # this runs

name = ""
print(bool(name))   # False (empty string)
print(bool("hi"))   # True
print(bool(0))      # False
print(bool(42))     # True

# A common pattern: provide a fallback for an empty value
username = "" or "guest"
print(username)     # guest
```

## Conditional (Ternary) Expressions

When you want to choose between two values, a one-line conditional expression is concise: `value_if_true if condition else value_if_false`.

```python
age = 20

status = "adult" if age >= 18 else "minor"
print(status)   # adult

# Useful inside an f-string
score = 72
print(f"Result: {'pass' if score >= 50 else 'fail'}")
```

## Nested Loops

A loop inside another loop is useful for combinations, grids, and tables. Be mindful: if the outer loop runs N times and the inner runs M times, the body runs N × M times.

```python
# Multiplication table (1 to 3)
for row in range(1, 4):
    for col in range(1, 4):
        print(f"{row * col:>3}", end="")
    print()

print("---")

# All pairs from two lists
colors = ["red", "blue"]
sizes = ["S", "L"]
for color in colors:
    for size in sizes:
        print(color, size)
```

## Writing Your Own Module

Any `.py` file is a module. You can `import` it to reuse its functions. Below we create a small module file, then import and use it. (Normally you would write the file yourself in an editor.)

```python
# Create a module file called mymath.py
with open("mymath.py", "w") as f:
    f.write("def square(n):\n")
    f.write("    return n * n\n\n")
    f.write("def cube(n):\n")
    f.write("    return n * n * n\n")

# Now import and use it
import mymath

print(mymath.square(5))   # 25
print(mymath.cube(3))     # 27
```

## Reading Structured Data from a File

Files often hold one record per line. A common task is to read each line, split it into fields, and build a list of dictionaries you can work with.

```python
# First, write some sample data (name,score per line)
with open("scores.txt", "w") as f:
    f.write("Alice,85\n")
    f.write("Bob,92\n")
    f.write("Carol,78\n")

# Read it back into structured data
records = []
with open("scores.txt", "r") as f:
    for line in f:
        name, score = line.strip().split(",")
        records.append({"name": name, "score": int(score)})

print(records)

# Now it is easy to work with
best = max(records, key=lambda r: r["score"])
print("Top scorer:", best["name"])
```

## Handling Errors More Carefully

You can catch different errors in different ways, and `raise` an error yourself when an input does not make sense.

```python
def safe_divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None

print(safe_divide(10, 2))   # 5.0
print(safe_divide(10, 0))   # prints message, returns None

# Raising your own error for invalid input
def set_age(age):
    if age < 0:
        raise ValueError("age cannot be negative")
    return age

try:
    set_age(-5)
except ValueError as error:
    print("Error:", error)
```

## Putting It Together: Word Frequency Counter

This program ties together strings, loops, dictionaries, sorting, and formatting — a small but realistic task.

```python
text = """
the quick brown fox jumps over the lazy dog
the dog was not amused the fox ran away
"""

# Clean and split into words
words = text.lower().split()

# Count each word
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

# Sort by count, highest first
ranked = sorted(counts.items(), key=lambda pair: pair[1], reverse=True)

print("Most common words:")
for word, count in ranked[:5]:
    print(f"{word:<8} {count}")
```

## Common Mistakes

- Forgetting that slices exclude the `stop` index: `s[0:3]` gives 3 characters, not 4.
- Expecting string methods to change the string in place — they return a new string, so assign the result.
- Assuming `alias = my_list` makes a copy. It does not; use `.copy()`.
- Putting a parameter with a default before one without a default in a function definition.
- Confusing `sorted(x)` (returns a new list) with `x.sort()` (changes `x`, returns `None`).
- Using `d[key]` for a key that might be missing instead of `d.get(key, default)`.

## Cheat Sheet

| Concept | Example |
| --- | --- |
| Slice | `text[1:4]`, `text[::-1]` |
| Strip / split / join | `s.strip()`, `s.split(",")`, `",".join(parts)` |
| Format number | `f"{x:.2f}"`, `f"{x:,}"` |
| Enumerate | `for i, v in enumerate(items):` |
| Zip | `for a, b in zip(xs, ys):` |
| Sort by key | `sorted(items, key=len, reverse=True)` |
| Comprehension | `[n*n for n in nums if n > 0]` |
| Dict default | `d.get(key, fallback)` |
| Multiple return | `return low, high` |
| Ternary | `"yes" if cond else "no"` |
| Raise error | `raise ValueError("message")` |
