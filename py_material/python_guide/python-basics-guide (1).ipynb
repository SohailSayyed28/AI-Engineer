## Comments

Comments are notes in your code. Python ignores them when running the program. Use `#` for a single-line comment.

````python
# This is a single-line comment
print("Comments help explain code")

"""
This is a multi-line comment.
It can span several lines.
"""
print("Done")
````

**Step by step:**

1. Line 1 starts with `#`, so Python skips it completely — it is only for human readers.
2. The first `print` runs normally; the comment above it changed nothing.

## Numbers and Math

Python can do arithmetic with the usual operators.

````python
a = 10
b = 3

print(a + b)   # Addition: 13
print(a - b)   # Subtraction: 7
print(a * b)   # Multiplication: 30
print(a / b)   # Division: 3.333...
print(a // b)  # Floor division: 3
print(a % b)   # Remainder (modulo): 1
print(a ** b)  # Exponent: 1000
````

**Step by step:**

1. We store `10` and `3` first, then use them in every calculation — change them once at the top and all results change.
2. `+`, `-`, `*` work as expected: 13, 7, 30.
3. `/` is **true division** and always gives a float: `3.333...` — even `10 / 2` gives `5.0`.
4. `//` is **floor division**: divide, then drop the decimal part → `3`.
5. `%` (modulo) gives the **remainder** of the division: 10 ÷ 3 = 3 remainder **1**. Very useful for "is this number even?" (`n % 2 == 0`).
6. `**` is power: `10 ** 3` = 10 × 10 × 10 = 1000.

## Strings

Strings hold text. You can use single or double quotes. The most convenient way to insert variables into text is an **f-string**.

````python
greeting = "Hello"
name = "Alice"

# Joining strings
print(greeting + ", " + name + "!")

# f-string (recommended)
age = 25
print(f"{name} is {age} years old.")
````

**Step by step:**

1. `greeting + ", " + name + "!"` glues strings together with `+`. It works, but you must manage every space and comma yourself.
2. The f-string does the same job more readably: the `f` before the quote switches on **replacement mode**.
3. Inside an f-string, anything in `{curly braces}` is evaluated as Python and the result is inserted into the text — `{name}` becomes `Alice`, `{age}` becomes `25`.

**Try it:** put a calculation inside the braces: `f"{name} will be {age + 5} in five years."`

````python
text = "Hello, World"

print(text.upper())                      # HELLO, WORLD
print(text.lower())                      # hello, world
print(text.replace("World", "Python"))   # Hello, Python
print(len(text))                         # 12
print(text.split(","))                   # ['Hello', ' World']
````

**Step by step:**

1. A **method** is a function that belongs to a value — you call it with a dot: `text.upper()`.
2. `upper()` and `lower()` return a **new** string; `text` itself is unchanged (print `text` afterwards to confirm).
3. `replace(old, new)` swaps every occurrence of `old`.
4. `len(text)` is a regular function, not a method — it counts characters, including spaces and punctuation.
5. `split(",")` cuts the string at every comma and returns a **list** of the pieces — notice the leading space in `' World'` because we only split on the comma.

## Type Conversion

Convert between types using built-in functions like `int()`, `float()`, and `str()`.

````python
number = int("42")     # string to int
price = float("3.99")  # string to float
text = str(100)        # int to string

print(number + 1)
print(price)
print(text)
````

**Step by step:**

1. `int("42")` reads the text `"42"` and produces the number `42` — now `number + 1` is real math, giving `43`.
2. `float("3.99")` does the same for decimals.
3. `str(100)` goes the other way: number → text, which you need before joining it to other strings with `+`.
4. Conversion can fail: `int("abc")` has no sensible answer and raises a `ValueError`. You'll learn to handle that in the errors section.

## Comparisons and Booleans

Comparisons return `True` or `False`.

| Operator | Meaning |
| --- | --- |
| `==` | Equal to |
| `!=` | Not equal to |
| `>` | Greater than |
| `<` | Less than |
| `>=` | Greater than or equal |
| `<=` | Less than or equal |

````python
print(5 > 3)    # True
print(5 == 5)   # True
print(5 != 3)   # True
print(2 >= 4)   # False
````

**Step by step:**

1. Each comparison is a question, and Python answers with the value `True` or `False` — a `bool` you can store in a variable like any other value.
2. `5 > 3` → is 5 greater than 3? `True`.
3. `5 == 5` → **double** equals asks "are these equal?". Single `=` would try to *assign* and fail here. This mix-up is the classic beginner bug.
4. `2 >= 4` → False. These True/False answers are exactly what `if` statements consume next.

## Conditions (if / elif / else)

Conditions let your program make decisions.

**Important:** Python uses indentation (spaces) to group code. Use four spaces for each level. Do not mix tabs and spaces. Also remember the colon `:` at the end of the line.

````python
age = 18

if age >= 18:
    print("You are an adult.")
elif age >= 13:
    print("You are a teenager.")
else:
    print("You are a child.")
````

**Step by step:**

1. Python evaluates `age >= 18` → with `age = 18` that's `True`, so the first block runs and **everything else is skipped**.
2. Order matters: conditions are checked top to bottom and only the *first* true one wins. If the `elif` came first, an 18-year-old would be called a teenager.
3. The indented lines (4 spaces) are the "body" — they belong to the condition above them. The colon `:` at the end of each condition line is required.

**Try it:** set `age = 15`, then `age = 8`, and predict the output before running.

````python
age = 20
has_ticket = True

if age >= 18 and has_ticket:
    print("You may enter.")

if age < 5 or age > 65:
    print("You get a discount.")

if not has_ticket:
    print("Please buy a ticket.")
````

**Step by step:**

1. `and` requires **both** sides to be true: `age >= 18 and has_ticket` → `True and True` → `True`, so the first message prints.
2. `or` needs **at least one** side: `age < 5 or age > 65` → `False or False` → `False`, so no discount message.
3. `not` flips a boolean: `not has_ticket` → `not True` → `False`, so no "buy a ticket" message.

**Try it:** set `age = 70` and `has_ticket = False`, predict which of the three messages print, then run.

## Loops

### for Loop

A `for` loop repeats over a sequence. `range(start, stop, step)` produces a sequence of numbers.

````python
for i in range(5):
    print(i)   # 0, 1, 2, 3, 4

print("---")

for i in range(0, 10, 2):
    print(i)   # 0, 2, 4, 6, 8

print("---")

fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
````

**Step by step:**

1. `range(5)` produces the numbers 0, 1, 2, 3, 4 — it starts at 0 and **stops before** 5. Five numbers, but never the 5 itself.
2. Each pass through the loop, `i` takes the next value and the indented body runs again.
3. `range(0, 10, 2)` means start at 0, stop before 10, jump by 2 → 0, 2, 4, 6, 8.
4. The third loop shows you don't need `range` at all when you have a list — `for fruit in fruits` visits each item directly. Prefer this form; it reads like English.

### while Loop

A `while` loop repeats as long as a condition is true. Make sure the condition will eventually become false, or the loop runs forever.

````python
count = 0
while count < 5:
    print(count)
    count = count + 1
````

**Step by step:**

1. Before the loop: `count = 0`.
2. Python checks the condition: `0 < 5`? Yes → run the body: print 0, then set `count` to 1.
3. Check again: `1 < 5`? Yes → print 1, `count` becomes 2 ... and so on.
4. When `count` reaches 5, the check `5 < 5` is False and the loop ends.
5. The line `count = count + 1` is what moves the loop forward. Delete it (mentally!) and the condition never becomes false — an **infinite loop**. That's why `while` needs more care than `for`.

### break and continue

- `break` stops the loop early.
- `continue` skips to the next iteration.

````python
for i in range(10):
    if i == 5:
        break       # stop the loop
    print(i)

print("---")

for i in range(5):
    if i == 2:
        continue    # skip this iteration
    print(i)
````

**Step by step:**

1. First loop: prints 0, 1, 2, 3, 4. When `i` becomes 5, `break` fires and the **whole loop stops** — 5 through 9 are never printed.
2. Second loop: when `i == 2`, `continue` says "skip the rest of the body **for this round only**" — so 2 is not printed, but the loop keeps going: 0, 1, 3, 4.
3. Remember the difference: `break` = leave the loop entirely; `continue` = skip to the next iteration.

## Lists

A list stores multiple values in order. Indexing starts at 0, and `-1` is the last item.

````python
fruits = ["apple", "banana", "cherry"]

print(fruits[0])    # apple
print(fruits[-1])   # cherry

fruits[1] = "blueberry"   # change an item
fruits.append("orange")   # add to the end
fruits.insert(0, "mango") # add at a position
fruits.remove("apple")    # remove by value
last = fruits.pop()       # remove and return the last item

print(fruits)
print("removed:", last)
````

**Step by step:**

1. `fruits[0]` → `"apple"`: positions start at **0**, so `[0]` is the first item. `fruits[-1]` counts from the end → `"cherry"`.
2. `fruits[1] = "blueberry"` replaces the item at position 1 — lists can be changed in place.
3. `append` adds to the end; `insert(0, "mango")` pushes an item in at position 0 (everything shifts right).
4. `remove("apple")` searches for the **value** and deletes the first match.
5. `pop()` removes the **last** item *and hands it back*, which is why we could store it in `last`.
6. Follow the prints: the final list is the result of all five changes in order. Work through it on paper if the output surprises you.

````python
numbers = [3, 1, 4, 1, 5, 9]

print(len(numbers))   # 6
print(sum(numbers))   # 23
print(min(numbers))   # 1
print(max(numbers))   # 9
numbers.sort()
print(numbers)        # [1, 1, 3, 4, 5, 9]

# Slicing
nums = [0, 1, 2, 3, 4, 5]
print(nums[1:4])   # [1, 2, 3]
print(nums[:3])    # [0, 1, 2]
print(nums[3:])    # [3, 4, 5]
````

**Step by step:**

1. `len`, `sum`, `min`, `max` are built-in functions that work on any list of numbers — no loop needed.
2. `numbers.sort()` rearranges the list **in place** (it returns nothing — a common surprise).
3. Slicing works exactly like string slicing: `nums[1:4]` takes positions 1, 2, 3 — the stop position **4 is excluded**.
4. Leaving out a side means "from the start" (`[:3]`) or "to the end" (`[3:]`). The original list is never changed by a slice — you get a new list.

## Tuples

A tuple is like a list but cannot be changed after creation. Use tuples for fixed groups of values.

````python
point = (3, 4)
print(point[0])   # 3
print(point[1])   # 4
````

**Step by step:**

1. Parentheses `(3, 4)` create a tuple; indexing works exactly like a list: `point[0]` → 3.
2. The difference: tuples are **immutable**. Try adding `point[0] = 99` to the cell — you'll get a `TypeError`.
3. Use a tuple when the values belong together and shouldn't change: a coordinate, an RGB color, a (name, score) pair.

## Dictionaries

A dictionary stores data as key and value pairs.

````python
person = {
    "name": "Alice",
    "age": 25,
    "city": "London"
}

print(person["name"])      # Alice
print(person.get("age"))   # 25

person["email"] = "alice@example.com"   # add
person["age"] = 26                       # change

for key, value in person.items():
    print(f"{key}: {value}")
````

**Step by step:**

1. The dictionary maps **keys** (`"name"`, `"age"`, `"city"`) to **values**. Look up by key: `person["name"]` → `"Alice"`.
2. `person.get("age")` does the same lookup, but if the key were missing it would return `None` instead of crashing — safer for keys that might not exist.
3. Assigning to a **new** key (`"email"`) adds an entry; assigning to an **existing** key (`"age"`) replaces its value.
4. `person.items()` yields `(key, value)` pairs, which the loop unpacks into two variables at once — this is *the* standard way to walk a dictionary.

## Sets

A set stores unique values with no duplicates. This makes sets useful for removing duplicates from a list.

````python
numbers = {1, 2, 2, 3, 3, 3}
print(numbers)   # {1, 2, 3}

items = [1, 1, 2, 3, 3]
unique = set(items)
print(unique)    # {1, 2, 3}
````

**Step by step:**

1. Even though we wrote `2` twice and `3` three times, the set keeps each value **once** — duplicates are silently dropped.
2. `set(items)` converts a list into a set, which is the one-line idiom for removing duplicates.
3. Notice sets have **no order** — don't rely on the printed order, and you can't index them with `[0]`.

## Functions

A function is a reusable block of code. It can take parameters and return values, and parameters can have default values.

````python
def greet(name):
    print(f"Hello, {name}!")

greet("Alice")

def add(a, b):
    return a + b

result = add(3, 4)
print(result)   # 7

def greet_default(name="friend"):
    print(f"Hello, {name}!")

greet_default()          # Hello, friend!
greet_default("Alice")   # Hello, Alice!
````

**Step by step:**

1. `def greet(name):` **defines** the function — nothing runs yet. `name` is a placeholder (a *parameter*) that will receive a value later.
2. `greet("Alice")` **calls** it: `"Alice"` is copied into `name` and the body runs.
3. `add` uses `return` to **send a value back** to the caller — that's why we can store the answer in `result`. `greet` only prints; it returns nothing (`None`).
4. `greet_default(name="friend")` gives the parameter a **default**: call it with no argument and the default is used; pass one and it's overridden.

**Try it:** write `def multiply(a, b=2):` and call it both ways.

## Handling Errors

Errors happen. You can handle them gracefully with `try` and `except` so the program does not crash. Change `user_value` to `"abc"` and re-run to see the error path.

````python
user_value = "42"  # imagine this came from input()

try:
    number = int(user_value)
    print(f"You entered {number}.")
except ValueError:
    print("That was not a valid number.")
````

**Step by step:**

1. Python **tries** the indented block under `try`.
2. With `user_value = "42"`, `int("42")` succeeds, the success message prints, and `except` is skipped entirely.
3. Change `user_value` to `"abc"`: now `int("abc")` raises a `ValueError`, Python **jumps immediately** to the matching `except` block, and the friendly message prints instead of a crash.
4. The `except ValueError:` line names exactly which error we expect — other kinds of errors would still crash, which is good: you only silence what you understand.

## Working with Files

The `with` block automatically closes the file when done.

| Mode | Meaning |
| --- | --- |
| `"r"` | Read (default) |
| `"w"` | Write (overwrites the file) |
| `"a"` | Append (adds to the end) |

````python
# Write to a file
with open("notes.txt", "w") as file:
    file.write("First line\n")
    file.write("Second line\n")

# Read the whole file
with open("notes.txt", "r") as file:
    print(file.read())

# Read line by line
with open("notes.txt", "r") as file:
    for line in file:
        print(line.strip())
````

**Step by step:**

1. `open("notes.txt", "w")` creates the file (or wipes it if it exists — careful with `"w"`); the `with` block guarantees it is closed afterwards, even on error.
2. `write` does **not** add newlines for you — that's why each string ends with `\n`.
3. The second block reopens the file in read mode and `file.read()` returns the whole content as one string.
4. The third block shows the more memory-friendly way: looping gives one line at a time. Each line still carries its `\n`, so `line.strip()` trims it before printing.

## Importing Modules

Python includes many modules. Use `import` to load them. Install extra packages with `pip install <package>`.

````python
import math

print(math.sqrt(16))   # 4.0
print(math.pi)         # 3.141592...

import random

print(random.randint(1, 6))   # a random number from 1 to 6
````

**Step by step:**

1. `import math` loads the module once; after that everything in it is available as `math.something`.
2. `math.sqrt(16)` → `4.0` (a float), `math.pi` is a ready-made constant.
3. `random.randint(1, 6)` returns a random integer from 1 to 6 **inclusive** — run the cell several times and watch the value change; it's a dice roll.
4. The pattern to remember: `import module`, then `module.function(...)`.

## A Small Complete Program

This program summarizes a list of numbers. In a script you might collect these from the user with `input()`; here we use a fixed list so the cell runs directly.

````python
numbers = [12.0, 7.5, 9.0, 20.0, 3.5]

if numbers:
    print(f"Count: {len(numbers)}")
    print(f"Sum: {sum(numbers)}")
    print(f"Average: {sum(numbers) / len(numbers)}")
    print(f"Largest: {max(numbers)}")
    print(f"Smallest: {min(numbers)}")
else:
    print("No numbers were entered.")
````

**Step by step:**

1. `if numbers:` uses the fact that an **empty list counts as False** — so this line asks "is there any data?" and protects the math below from an empty list (dividing by `len([])` would crash).
2. Each print combines an f-string with one built-in: `len` (how many), `sum`, `max`, `min`.
3. The average is just `sum / len` — no special function needed.

**Try it:** set `numbers = []` and re-run to see the `else` branch; then add your own numbers.
